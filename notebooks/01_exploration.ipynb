## Exploratory Data Analysis

```python
# Load sample spectrum and check dimensions
import pickle

# Create sample spectrum for demonstration
sample_spectrum = np.linspace(0, 1, 1800)
wavenumber_axis = np.linspace(400, 4000, 1800)

print(f"Spectrum shape: {sample_spectrum.shape}")
print(f"Wavenumber range: {wavenumber_axis[0]:.0f}-{wavenumber_axis[-1]:.0f} cm^-1")
print(f"Wavenumber resolution: {wavenumber_axis[1]-wavenumber_axis[0]:.2f} cm^-1")
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 4))
plt.plot(wavenumber_axis, sample_spectrum, linewidth=0.8)
plt.xlabel("Wavenumber (cm⁻¹)")
plt.ylabel("Absorbance")
plt.title("Sample IR Spectrum")
plt.grid(alpha=0.3)
plt.show()
```
