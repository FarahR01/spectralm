```python
from spectralm.physics.isotope_shifts import IsotopeShiftCorrector

corrector = IsotopeShiftCorrector()

# Example shift calculation
freq_c12 = 1200.0  # cm^-1 for 12C=O
shifted = corrector.predict_shift(
    freq_primary=freq_c12,
    atom_pair="C-O",
    primary_isotope=12,
    target_isotope=13,
)

print(f"Isotope shift example:")
print(f"  Original C12=O peak: {freq_c12:.1f} cm⁻¹")
print(f"  With C13 substitution: {shifted:.1f} cm⁻¹")
print(f"  Shift magnitude: {freq_c12 - shifted:.1f} cm⁻¹")
shift_pct = 100 * (freq_c12 - shifted) / freq_c12
print(f"  Shift percentage: {shift_pct:.2f}%")
```
